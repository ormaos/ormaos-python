# A notebook that records one cell with `with ormaos.record():`

```python
def parse(text):
    try:
        return int(text)
    except ValueError:
        return 0  # not a number counts as zero


items = ["1", "x", "3"]
```

```python
import ormaos

with ormaos.record() as recording:
    print("total", sum(parse(item) for item in items))
print("recorded", recording.name)
```

```python
print("after", parse("7"))
```
